**CSV_data2 represents the following**

**4. Occupation_Type**

* This column provides the type of occupation of the participant.

**5. Chronotype**

* This column provides the type of sleep-wake pattern of the participant. Chronotype by definition is an individual's body's natural, genetically predetermined preference for when to sleep and wake up during a 24-hour day.

**6. Bedtime_Phone_Minutes**

* This column provides the average number of minutes the participant was on their phone during bedtime.

**7. Primary_Bedtime_App**

* This column provides information about the participant's primary bedtime app - what they tend to use the most at this time of their day.

**8. Screen_Brightness_Pct**

* This column provides information about the percentage of screen brightness with which the participant was operating their phone at bedtime.

**9. Blue_Light_Filter_Active**

* This column indicates whether the participant had their blue light filter active during their time of usage. Blue light filter is an active display setting that reduces the amount of high-energy short-wavelength bluelight emitted from a screen.

**10. Caffeine_post_5pm_mg**

* This column takes into consideration the effect of caffeine consumption a participant had in the evening hours.

**11. Physical_Activity_Min**

* This column monitors how many minutes of physical activity the participant engaged in during the day.

**12. Sleep_Latency_Min**

* This column measures the amount of time it takes the participant to transition from full wakeness to the very first stage of sleep after they turn off the lights and try to rest.

**13. Total_Sleep_Hours**

* This column indicates the average total number of sleep hours the participant had each night.

**14. Deep_Sleep_Pct**

* This column indicates the percentage of deep sleep the participant experienced on average at night. Deep sleep is the most physically restorative stage of sleep.

**15. REM_Sleep_Pct**

* This column indicates the percent of REM sleep the participant experienced on average at night. REM sleep is the active phase of the sleep cycle when the most vivid dreams occur.

**16. Morning_Alarm_Snoozes**

* This column indicates on average how many times the participant hit the snooze button on the alarm clock in the morning to wake up.

**17. Next_Day_Fatigue_Score**

* This column indicates the participant's next day fatigue score. This score is a predictive metric used in digital health and sleep research to estimate how tired or low-energy a person will feel on the following day based on current biometric data.

**18. Sleep_Debt_Category**

* This column uses a descriptive adjective to describe the severity of sleep debt of the participant.

```python
df7 = '/content/01 Retail Inventory.xlsx'
read_df7 = pd.read_excel(df7)
read_df7.head(11)
```

**ExcelData7 represents the following**

This dataset provides information about the retail inventory of a store.

**Columns**

**1. Product ID**

* This column tells the a unique alphanumeric code assigned to a specific item to track it, manage inventory, or look it up in a database.

**2. Product Name**

* This column provides the product name. The product name is the specific title given to an item or data package to identify it easily.

**3. Category**

* This column provides information about which category the product falls under. The category refers to a high-level grouping of products that share similar characteristics, uses, or customer target markets.

**4. Supplier Name**

* This column shares the supplier name of the product. A supplier name is the official business name of the manufacturer, vendor, or distributor that provides goods or raw materials to another company

**5. Unit Cost**

* This column shares the unit cost of the product. Unit cost is the total amount of money it costs a company to produce, buy, or source a single individual unit of a product.

**6. Retail Price**

* This column conveys the retail price of the product. Retail price is the final amount of money a consumer pays to purchase a single unit of a product at a store or online.

**7. Opening Stock**

* This column refers to what is also known as "beginning stock." It is the total monetary value or quantity of products that a business has on hand and available for sale at the very start of a specific accounting period (such as a new month, quarter, or fiscal year).

**8. Units Sold**

* This column refers to the total quantity of individual items purchased by customers during a specific timeframe (e.g., daily, monthly, or annually). It tracks product demand and is directly used to calculate total sales revenue.

**9. Closing Stock**

* This column provides information regarding the total value or physical quantity of products remaining on hand and available for sale at the very end of an accounting period. The closing stock of one period automatically becomes the opening stock of the next period.

**10. Reorder Point**

* This column refers to the specific minimum inventory level that triggers the need to buy more stock from a supplier. It is calculated based on how fast the product sells (demand) and how long it takes the supplier to deliver it (lead time) to ensure the store never runs out of stock.

**11. Stock Status**

* This column refers to the current availability of an item in an inventory system.
